```python
import numpy as np 
```

```python
def shuffle_and_split(data,test_ratio):
    np.random.seed(42) # set the seed for reproducibility 
    shuffled_indices = np.random.permutation(len(data))#this return suffled indices
    test_set_size = int(len(data) * test_ratio )
    test_indices = shuffled_indices[:test_set_size]
    train_indices = shuffled_indices[test_set_size:]
    return data.iloc[train_indices],data.iloc[test_indices]
```

```python
import pandas as pd 
```

```python
data = pd.read_csv("housing.csv")
```

```python
data 
```

```python
data["income_category"] = pd.cut(data['median_income'],bins = [0,1.5,3.0,4.5,6.0,np.inf],labels=[1,2,3,4,5]  )
```

```python
import matplotlib.pyplot as plt 
```

```python
data["income_category"].value_counts().sort_index().plot.bar(rot = 0 , grid = True )
plt.title("Income Category Distribution ")
plt.xlabel('income Category')
plt.ylabel('number of instances')
plt.show()
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
split = StratifiedShuffleSplit(n_splits = 1 ,test_size = 0.2,random_state = 42)
```

```python
for train_index, test_index in split.split(data,data['income_category']):
    strat_test = data.loc[test_index]
    strat_train = data.loc[train_index]
```

```python
strat_train
```

```python
strat_train["income_category"].value_counts().sort_index().plot.bar(rot = 0 , grid = True )
plt.title("Income Category Distribution ")
plt.xlabel('income Category')
plt.ylabel('number of instances')
plt.show()
```

```python
strat_test["income_category"].value_counts().sort_index().plot.bar(rot = 0 , grid = True )
plt.title("Income Category Distribution ")
plt.xlabel('income Category')
plt.ylabel('number of instances')
plt.show()
```

# 📊 Measuring Errors (MAE & RMSE)

After training a **Regression Model**, we need to evaluate how accurate its predictions are.

Unlike classification problems (where we use **Accuracy**), regression problems predict **continuous numerical values** (such as house prices).

Therefore, we use **error metrics** to measure how far the predicted values are from the actual values.

Two of the most commonly used metrics are:

- **MAE (Mean Absolute Error)**
- **RMSE (Root Mean Squared Error)**

---

# 🎯 Why Measure Errors?

Suppose the actual price of a house is:

**₹80 Lakhs**

The model predicts:

**₹75 Lakhs**

The prediction is close but not perfect.

The **error** is:

```text
Error = Actual − Predicted
      = 80 − 75
      = 5 Lakhs
```

Error metrics summarize these differences over all predictions.

---

# 1️⃣ Mean Absolute Error (MAE)

**MAE (Mean Absolute Error)** measures the **average absolute difference** between the actual values and the predicted values.

The word **absolute** means we ignore positive and negative signs.

---

## 📐 Formula

```text
MAE = (1/n) × Σ |Actual − Predicted|
```

Where:

- **n** = Number of observations
- **Σ** = Sum of all errors
- **| |** = Absolute value

---

## 📝 Example

Suppose:

| Actual | Predicted |
|---------|-----------|
| 100 | 90 |
| 120 | 125 |
| 150 | 140 |

### Step 1: Calculate Absolute Errors

| Actual | Predicted | Absolute Error |
|---------|-----------|----------------|
|100|90|10|
|120|125|5|
|150|140|10|

Total Error:

```text
10 + 5 + 10 = 25
```

Number of observations:

```text
3
```

Therefore,

```text
MAE = 25 / 3
     = 8.33
```

**Answer: MAE = 8.33**

---

# 📌 Properties of MAE

- Easy to understand.
- Treats all errors equally.
- Less sensitive to large errors.
- Measured in the same units as the target variable.

---

# 🧠 MAE Uses L1 Norm (Manhattan Distance)

MAE is based on the **L1 Norm**, also called the **Manhattan Distance**.

Imagine walking in a city with grid-like roads.

You cannot move diagonally.

You move only:

- Left
- Right
- Up
- Down

The total distance traveled is the Manhattan Distance.

Similarly, MAE simply adds the absolute errors.

---

# 2️⃣ Root Mean Squared Error (RMSE)

**RMSE (Root Mean Squared Error)** measures the square root of the average squared errors.

Instead of taking absolute values, RMSE squares every error.

This makes large errors much more significant.

---

## 📐 Formula

```text
RMSE = √[(1/n) × Σ (Actual − Predicted)²]
```

---

## 📝 Example

Using the same data:

| Actual | Predicted |
|---------|-----------|
|100|90|
|120|125|
|150|140|

### Step 1: Square the Errors

| Actual | Predicted | Squared Error |
|---------|-----------|---------------|
|100|90|100|
|120|125|25|
|150|140|100|

Sum of squared errors:

```text
100 + 25 + 100 = 225
```

Average:

```text
225 / 3 = 75
```

Square root:

```text
√75 ≈ 8.66
```

**Answer: RMSE = 8.66**

---

# 📌 Properties of RMSE

- Penalizes large errors more heavily.
- Sensitive to outliers.
- Commonly used in regression tasks.
- Measured in the same units as the target variable.

---

# 🧠 RMSE Uses L2 Norm (Euclidean Distance)

RMSE is based on the **L2 Norm**, also known as the **Euclidean Distance**.

Euclidean distance measures the shortest straight-line distance between two points.

Since errors are squared before averaging, larger mistakes have a much bigger impact.

---

# ⚖️ MAE vs RMSE

| Feature | MAE | RMSE |
|----------|-----|------|
| Full Form | Mean Absolute Error | Root Mean Squared Error |
| Formula | Average of absolute errors | Square root of average squared errors |
| Large Errors | Treated equally | Penalized more |
| Sensitive to Outliers | ❌ Less | ✅ More |
| Norm Used | L1 (Manhattan) | L2 (Euclidean) |
| Interpretation | Easier | More sensitive |

---

# 📊 Example Comparison

Suppose one prediction is extremely wrong.

| Actual | Predicted |
|---------|-----------|
|100|102|
|105|104|
|110|250|

MAE increases moderately.

RMSE increases dramatically because the large error is squared.

This is why RMSE is preferred when large prediction errors are especially undesirable.

---

# 🐍 Calculating MAE and RMSE in Scikit-Learn

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import root_mean_squared_error

y_true = [100, 120, 150]
y_pred = [90, 125, 140]

mae = mean_absolute_error(y_true, y_pred)
rmse = root_mean_squared_error(y_true, y_pred)

print("MAE :", mae)
print("RMSE:", rmse)
```

> **Note:** In older versions of Scikit-Learn, RMSE is calculated as:

```python
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(mean_squared_error(y_true, y_pred))
```

---

# 📋 Summary

- Regression models are evaluated using **error metrics**, not accuracy.
- **MAE** calculates the average absolute difference between actual and predicted values.
- **RMSE** calculates the square root of the average squared differences.
- MAE treats all errors equally.
- RMSE gives more importance to large errors.
- A **lower MAE or RMSE** indicates a better-performing model.

---

# 🧠 Quick Exam Notes

### MAE (Mean Absolute Error)

> Average of the absolute differences between actual and predicted values.

### Formula

```text
MAE = (1/n) × Σ |Actual − Predicted|
```

### RMSE (Root Mean Squared Error)

> Square root of the average squared differences between actual and predicted values.

### Formula

```text
RMSE = √[(1/n) × Σ (Actual − Predicted)²]
```

### MAE

- Uses **L1 (Manhattan) Norm**
- Treats all errors equally
- Less affected by outliers

### RMSE

- Uses **L2 (Euclidean) Norm**
- Penalizes larger errors more
- More sensitive to outliers

### Rule of Thumb

- **Use MAE** when all errors are equally important.
- **Use RMSE** when larger errors should be penalized more heavily.

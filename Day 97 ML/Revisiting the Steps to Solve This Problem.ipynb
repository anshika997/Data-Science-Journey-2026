# 🔄 Revisiting the Steps to Solve This Problem

Before building a Machine Learning model, it is important to understand the nature of the problem.

This helps us choose the correct Machine Learning technique and algorithm.

---

# 🏠 Understanding the Problem

Our goal is:

> **Predict the price of a house based on its features.**

The model will learn from historical housing data and estimate the price of a new house.

---

# ❓ Step 1: Is This Supervised or Unsupervised Learning?

Ask yourself:

> **Do we already know the correct answers (house prices) in our training data?**

✅ **Yes**

Our dataset contains:

- House features (inputs)
- Actual house prices (outputs)

Since the target values are already known, this is a **Supervised Learning** problem.

### Why?

The model learns from labeled examples.

Example:

| Size | Rooms | Location | House Price |
|------|-------|----------|-------------|
| 1200 sq ft | 2 | Gurgaon | ₹60 Lakhs |
| 1800 sq ft | 3 | Gurgaon | ₹95 Lakhs |

Here, the **House Price** is the known label (target).

---

# ❓ Step 2: Classification or Regression?

Now ask:

> **What are we trying to predict?**

Are we predicting:

- Categories (Yes/No, Cat/Dog)? ❌
- A numerical value? ✅

Our model predicts the **price of a house**, which is a continuous numerical value.

Examples:

- ₹45 Lakhs
- ₹72 Lakhs
- ₹1.25 Crores

Therefore, this is a **Regression** problem.

---

# ❓ Step 3: Univariate or Multivariate Regression?

The model uses **multiple input features**, such as:

- Median Income
- House Age
- Number of Rooms
- Population
- Latitude
- Longitude

to predict **one output**:

- House Price

Since there is **only one target (output) variable**, this is called a **Univariate Regression** problem.

> **Note:** "Univariate" refers to having **one target variable**, even though there are multiple input features.


# Scikit-Learn's Design Principles

Scikit-Learn provides a simple and consistent API that makes Machine Learning tools easier to learn and use.

## 1. Consistency

All objects follow a standard interface, making different Scikit-Learn tools easier to understand and use.

## 2. Estimators

An **estimator** is an object that learns information from data.

- Use `.fit()` to learn from data.
- In supervised learning, use `.fit(X, y)` to provide features (`X`) and labels (`y`).
- **Hyperparameters** are settings provided when creating an object.

Example:

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")
imputer.fit(data)
```

Here, `strategy="median"` is a hyperparameter.

## 3. Transformers

A **transformer** is an estimator that can transform data.

- `.transform()` applies a transformation after fitting.
- `.fit_transform()` learns the required information and transforms the data in one step.

Example:

```python
X_transformed = imputer.fit_transform(data)
```

## 4. Predictors

A **predictor** is an object that makes predictions using a trained model.

- `.predict()` generates predictions for new data.
- `.score()` evaluates model performance. The metric depends on the estimator—for example, accuracy for classifiers or \(R^2\) for many regressors.

Example:

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

predictions = model.predict(X_test)

score = model.score(X_test, y_test)
```

## 5. Inspection

Scikit-Learn makes it easy to inspect hyperparameters and learned parameters.

### Hyperparameters

Hyperparameters are usually accessible directly through their attribute names.

```python
imputer.strategy
```

### Learned Parameters

Learned parameters often have an underscore (`_`) at the end of their attribute names.

```python
imputer.statistics_
model.coef_
```

- `imputer.statistics_` contains the learned imputation values.
- `model.coef_` contains the learned coefficients of a linear regression model.

## 6. No Extra Classes

Scikit-Learn generally uses familiar data structures, such as NumPy arrays and Pandas DataFrames, for inputs and outputs.

This reduces the need to learn many custom data types.

## 7. Composition

Scikit-Learn allows multiple preprocessing and modeling steps to be combined into a **Pipeline**.

A pipeline applies the steps in sequence, helping keep preprocessing consistent during training and prediction.

Example:

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", LinearRegression())
])

pipeline.fit(X_train, y_train)
```

Here:
1. The imputer fills missing numerical values using the median.
2. The Linear Regression model learns from the transformed features.

**Note:** This example assumes that the input features are numerical.

## 8. Sensible Defaults

Many Scikit-Learn tools provide sensible default settings, allowing beginners to get started without configuring every parameter manually.

However, defaults may not be optimal for every dataset or problem.

# Working with Pandas DataFrames

Even if you provide a Pandas DataFrame as input, a transformer may return a NumPy array instead of a DataFrame.

Example:

```python
X = imputer.transform(housing_num)
```

To convert the resulting NumPy array back into a Pandas DataFrame:

```python
import pandas as pd

housing_tr = pd.DataFrame(
    X,
    columns=housing_num.columns,
    index=housing_num.index
)
```

Here:

- `X` contains the transformed numerical data.
- `columns=housing_num.columns` preserves the original column names.
- `index=housing_num.index` preserves the original row indices.
- `housing_tr` is the resulting Pandas DataFrame.

## Quick Revision Table

| Concept | Purpose | Common Method or Attribute |
|---|---|---|
| Estimator | Learns from data | `.fit()` |
| Transformer | Transforms data | `.transform()` |
| Combined transformation | Fits and transforms | `.fit_transform()` |
| Predictor | Makes predictions | `.predict()` |
| Model evaluation | Evaluates performance | `.score()` |
| Hyperparameters | Settings chosen before fitting | `strategy`, `fit_intercept` |
| Learned parameters | Values learned from data | `statistics_`, `coef_` |
| Pipeline | Combines multiple steps | `Pipeline([...])` |

## Key Takeaway

Scikit-Learn follows a consistent design:

- **`fit()`** → learn from data.
- **`transform()`** → transform data.
- **`predict()`** → make predictions.
- **`score()`** → evaluate performance.
- **`Pipeline`** → combine preprocessing and modeling steps.

```python
import pandas as pd

df = pd.read_excel("iris_train.xlsx", engine="openpyxl")


df.head()
```

```python
x = df.iloc[:,:-1]
```

```python
x
```

```python
y = df.iloc[:,-1]
```

```python
y
```

```python
model = RandomForestClassifier()
```

```python
model.fit(x.values,y.values)
```

```python
predict = model.predict([[5.4, 3.9, 1.3, 0.4]])
```

```python
predict
```

```python
datatest = pd.read_excel("iris_test.xlsx",engine = "openpyxl")
```

```python
x_test = datatest.iloc[:,:-1]
```

```python
predict = model.predict(x_test.values)
```

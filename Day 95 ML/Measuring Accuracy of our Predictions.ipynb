# 📊 Measuring Accuracy of Our Predictions

When we build a **Machine Learning classification model**, we need a way to measure how well it performs.

For example, suppose our model predicts:

- Spam or Not Spam
- Cat or Dog
- Apple or Orange
- Healthy or Diseased

After making predictions, we compare them with the actual answers to see how many predictions were correct.

One of the simplest evaluation metrics is **Accuracy**.

---

# 🎯 What is Accuracy?

**Accuracy** tells us **how often the model makes correct predictions**.

In simple words:

> **Accuracy is the percentage of predictions that the model got right.**

---

# 📐 Accuracy Formula

```text
Accuracy = (Number of Correct Predictions) / (Total Number of Predictions)
```

Or,

```text
Accuracy = Correct Predictions ÷ Total Predictions
```

---

# 📝 Example 1

Suppose a model predicts the results of **100 emails**.

Out of these:

- 90 predictions are correct ✅
- 10 predictions are incorrect ❌

Then,

```text
Accuracy = 90 / 100
         = 0.90
         = 90%
```

**Answer:** The model has **90% accuracy**.

---

# 📝 Example 2

A fruit classifier predicts **50 fruits**.

- Correct predictions = 47
- Incorrect predictions = 3

```text
Accuracy = 47 / 50
         = 0.94
         = 94%
```

The model's accuracy is **94%**.

---

# 📝 Example 3

A student answered **20 questions**.

- Correct = 18
- Wrong = 2

```text
Accuracy = 18 / 20
         = 0.90
         = 90%
```

This is the same idea used to evaluate Machine Learning models.

---

# 🤖 Measuring Accuracy in Scikit-Learn

Scikit-Learn provides the `accuracy_score()` function.

### Example

```python
from sklearn.metrics import accuracy_score

# Actual labels
y_true = ["cat", "dog", "cat", "dog"]

# Model predictions
y_pred = ["cat", "dog", "cat", "cat"]

accuracy = accuracy_score(y_true, y_pred)

print("Accuracy:", accuracy)
```

### Output

```text
Accuracy: 0.75
```

This means the model correctly predicted **75%** of the samples.

---

# 🧠 Understanding `accuracy_score()`

```python
accuracy_score(y_true, y_pred)
```

### Parameters

- `y_true` → Actual (correct) labels.
- `y_pred` → Labels predicted by the model.

### Returns

A value between **0 and 1**.

Example:

| Value | Meaning |
|--------|---------|
| 1.0 | 100% Accurate |
| 0.9 | 90% Accurate |
| 0.8 | 80% Accurate |
| 0.5 | 50% Accurate |

---

# 📊 Accuracy Calculation Example

| Actual | Predicted | Correct? |
|---------|-----------|----------|
| Cat | Cat | ✅ |
| Dog | Dog | ✅ |
| Cat | Dog | ❌ |
| Dog | Dog | ✅ |

Correct Predictions = **3**

Total Predictions = **4**

```text
Accuracy = 3 / 4
         = 0.75
         = 75%
```

---

# ⚠️ Is Accuracy Always the Best Metric?

Not always.

Imagine a dataset of **1000 emails**:

- 990 are **Not Spam**
- 10 are **Spam**

Suppose the model predicts **every email as "Not Spam"**.

Correct predictions = **990**

```text
Accuracy = 990 / 1000
         = 99%
```

The model has **99% accuracy**, but it failed to identify **all spam emails**.

So, although accuracy is high, the model is not very useful.

For such cases, we use other metrics like:

- Precision
- Recall
- F1-Score
- Confusion Matrix

(You'll learn these later.)

---

# 📋 Summary

- **Accuracy** measures how often a model makes correct predictions.
- It is calculated as:

```text
Accuracy = Correct Predictions / Total Predictions
```

- Accuracy is commonly used to evaluate **classification models**.
- In Scikit-Learn, use `accuracy_score()` to calculate accuracy.
- While accuracy is simple and intuitive, it may not always be the best metric for imbalanced datasets.

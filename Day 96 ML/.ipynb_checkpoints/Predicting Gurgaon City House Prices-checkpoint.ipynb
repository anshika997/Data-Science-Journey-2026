# 🏠 Predicting Gurgaon City House Prices

## Why Are We Building This?

**Gurgaon (Gurugram)** is one of India's fastest-growing cities and has experienced rapid real estate development over the past decade.

Due to:

- Proximity to Delhi
- Growing IT and corporate hubs
- Modern infrastructure
- Increasing demand for housing

property prices in Gurgaon change frequently and can be difficult to estimate accurately.

Machine Learning can help us predict house prices based on historical data and property features.

---

# 🎯 Objectives of the Project

We aim to build a Machine Learning model that can predict house prices.

The model can help:

### 🏡 Home Buyers

Estimate whether a property's price is fair before purchasing.

---

### 💰 Home Sellers

Set a reasonable selling price based on market trends.

---

### 🤝 Real Estate Agents

Provide better recommendations and improve negotiations.

---

### 📈 Investors

Analyze market trends and identify profitable investment opportunities.

---

# 🔍 Factors That Affect House Prices

House prices depend on several features, such as:

- 📍 Location
- 🏠 House Size
- 🛏️ Number of Bedrooms
- 🛁 Number of Bathrooms
- 🚗 Parking Availability
- 🌳 Nearby Amenities
- 🚇 Connectivity
- 🏢 Infrastructure
- 🏫 Schools and Hospitals
- 🛒 Shopping Centers

These features become the **input variables (features)** for our Machine Learning model.

---

# 📊 Dataset for This Project

Currently, we do not have a large, clean dataset containing Gurgaon house prices.

Therefore, we will use the **California Housing Dataset** as a substitute (proxy dataset).

Although the data is from California, it contains realistic housing information that helps us understand how house price prediction works.

Later, the same techniques can be applied to real Gurgaon housing data.

---

# 📋 Features in the California Housing Dataset

The dataset contains several useful features.

| Feature | Description |
|----------|-------------|
| **Median Income** | Average income of households in the area |
| **House Age** | Average age of houses |
| **Average Rooms** | Average number of rooms per house |
| **Average Bedrooms** | Average number of bedrooms |
| **Population** | Population living in the area |
| **Average Occupancy** | Average number of people per house |
| **Latitude** | Geographic latitude |
| **Longitude** | Geographic longitude |

---

# 🎯 Target Variable

The value we want to predict is:

**Median House Value**

This represents the estimated price of a house.

# 🛠️ Machine Learning Techniques Used

This project is a **Supervised Learning** problem because:

- We have input features (house characteristics).
- We have corresponding target values (house prices).

Since the target is a numerical value, this is a **Regression** problem.

Common regression algorithms include:

- Linear Regression
- Decision Tree Regressor
- Random Forest Regressor
- Gradient Boosting Regressor

We will start with **Linear Regression**, one of the simplest and most widely used regression algorithms.

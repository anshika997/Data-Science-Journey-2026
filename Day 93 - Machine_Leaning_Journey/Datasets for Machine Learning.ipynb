# 📂 Datasets for Machine Learning

Machine Learning models learn from **data**.

The quality of a Machine Learning model depends heavily on the quality of the dataset used for training and testing.

Without good data, even the best algorithms cannot produce accurate predictions.

---

# 🔹 What Is a Dataset?

A **dataset** is a collection of data organized in rows and columns.

Each row represents one observation (or sample), and each column represents a feature (attribute).

### Example

| Height (cm) | Weight (kg) | Gender |
|--------------|-------------|---------|
| 170 | 65 | Male |
| 160 | 55 | Female |
| 180 | 80 | Male |

Here:

- **Height** and **Weight** are **features**.
- **Gender** is the **label (target)**.

---

# 🎯 Why Are Datasets Important?

Datasets are used to:

- Train Machine Learning models
- Test model performance
- Compare different algorithms
- Discover patterns in data

The more relevant and high-quality the dataset, the better the model can learn.

# 🌸 The Iris Dataset

One of the most famous beginner datasets in Machine Learning is the **Iris Dataset**.

It is commonly used to learn:

- Classification
- Data visualization
- Model evaluation

---

## About the Iris Dataset

The dataset contains measurements of **three different species of Iris flowers**.

### Species

- Iris Setosa
- Iris Versicolor
- Iris Virginica

---

## Features

Each flower has four measurements:

| Feature | Description |
|----------|-------------|
| Sepal Length | Length of the sepal |
| Sepal Width | Width of the sepal |
| Petal Length | Length of the petal |
| Petal Width | Width of the petal |

---

## Target (Label)

The model predicts the flower species.

Example:

| Sepal Length | Sepal Width | Petal Length | Petal Width | Species |
|---------------|-------------|--------------|-------------|----------|
| 5.1 | 3.5 | 1.4 | 0.2 | Setosa |

# 🐍 Loading the Iris Dataset in Scikit-Learn

Scikit-Learn provides the Iris dataset directly.

```python
from sklearn.datasets import load_iris

iris = load_iris()

print(iris.data[:5])
print(iris.target[:5])
```
